```python
def analyze_file(filename):
    result = {'characters': 0, 'words': 0, 'lines': 0}
    
    try:
        with open(filename, 'r') as file:
            for line in file:
                result['lines'] += 1
                result['characters'] += len(line)
                result['words'] += len(line.split())
    except FileNotFoundError:
        print(f"File not found")
    
    return result
```

```python
def search_word(filename, word):
    count = 0
    
    try:
        with open(filename, 'r') as file:
            for line in file:
                count += line.split().count(word)
    except FileNotFoundError:
        print(f"The file '{filename}' was not found.")
    
    return count
```

## Test Case

```python
outcome = analyze_file('sam_file.txt')
print("File Analysis:", outcome)

word_count = search_word('sam_file.txt', 'name')
print("Word Count:", word_count)
```
